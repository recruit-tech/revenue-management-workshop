# Hotel pricing workshop: set the scenario

You manage a hotel with a fixed number of rooms. How should your offered price change with demand, the time remaining and the rooms still available? Every accepted sale permanently consumes one room.

Our goal is to maximise **expected revenue**, rather than sell every room at any price. The inputs below are invented teaching assumptions, not client observations. All prices are in arbitrary currency units.

Run this notebook from the repository root. It creates a new `workshop-run/` folder. Then open **02 → 03 inside that folder**. Existing workspaces are never overwritten.

```python
from pathlib import Path
import yaml
from python.synthetic import prepare_workspace

scenario_path = Path('scenarios/synthetic_demo.yaml')
with scenario_path.open(encoding='utf-8') as file:
    scenario = yaml.safe_load(file)
scenario
```
